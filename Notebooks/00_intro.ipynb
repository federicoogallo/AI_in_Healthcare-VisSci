```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from matplotlib.colors import LinearSegmentedColormap

sns.set_theme(style="whitegrid")

PURPLE_MAIN = "#caa5f1"
PURPLE_LIGHT = "#E9E0FF"

PURPLE_PALETTE = [
    "#E9E0FF",  
    "#E1D3FB",
    "#D7BFF6",
    "#D0B1F4",
    "#caa5f1"
]

DATA_PATH = Path("/Users/federico/Desktop/progetto VIS/0. intro/00_ai_job_trends.csv")
df = pd.read_csv(DATA_PATH)
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
industry_counts = df["Industry"].value_counts().sort_values()
cmap = LinearSegmentedColormap.from_list("my_purple", PURPLE_PALETTE)
purple_grad = [cmap(i / (len(industry_counts)-1)) for i in range(len(industry_counts))]
ax.barh(industry_counts.index, industry_counts.values, color=purple_grad)
ax.set_xlabel("Numero di lavori")
ax.set_ylabel("Settore")
ax.set_title("Distribuzione dei lavori per settore industriale")
plt.tight_layout()
plt.show()
```
